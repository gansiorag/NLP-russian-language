```python
# Программный поиск нужного dataset через Python:
from huggingface_hub import list_datasets

# ✅ sort без префикса, валидные значения: "likes", "trending", "lastModified", "createdAt"
datasets = list_datasets(
    language=["ru"],
    task_categories=["text-classification"],
    sort="likes",  # ✅ Без префикса "-"
    # direction=-1,  # ✅ Отдельный параметр для направления (если поддерживается)
    limit=30
)

datasets_list = list(datasets)
for ds in datasets_list:
    print(f"📦 {ds.id} — 👍 {ds.likes or 0} likes — 📅 {ds.description.replace('\n',  ' ').replace('  ',  ' ')}")
```
